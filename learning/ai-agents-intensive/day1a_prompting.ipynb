```python
from google.adk.agents import Agent
from google.adk.models.google_llm import Gemini
from google.adk.runners import InMemoryRunner
from google.adk.tools import google_search
from google.genai import types
import litellm

from duckduckgo_search import DDGS

print("✅ ADK components imported successfully.")
```

```python
retry_config=types.HttpRetryOptions(
    attempts=5,  # Maximum retry attempts
    exp_base=7,  # Delay multiplier
    initial_delay=1, # Initial delay before first retry (in seconds)
    http_status_codes=[429, 500, 503, 504] # Retry on these HTTP errors
)
```

```python
def local_web_search(query: str):
    """Searches the web for the given query."""
    # Example using a free library like duckduckgo_search
    results = DDGS().text(query, max_results=3)
    return str(results)
```

```python
import litellm
root_agent = Agent(
    name="helpful_assistant",
    model="ollama/llama3.2:3b", # replace llama3.2 with your specific downloaded model
    description="A simple agent that can answer general questions.",
    instruction="You are a helpful assistant. Use Google Search for current info or if unsure.",
    tools = [local_web_search],
)

print("Agent defined")
```

```python
runner = InMemoryRunner(agent=root_agent)
```

```python
response = await runner.run_debug(
    "What is Agent Development Kit from Google? What languages is the SDK available in?"
)
```

```python
from langgraph.prebuilt import create_react_agent
from langchain_ollama import ChatOllama
from langchain_community.tools import DuckDuckGoSearchRun

# Initialize
model = ChatOllama(model="llama3.2:3b")
tools = [DuckDuckGoSearchRun()]

# Create the agent (This version is now the standard)
agent_executor = create_react_agent(model, tools)

# Usage
response = agent_executor.invoke({"messages": [("user", "What is the current news regarding the Iran-Israel war? I want you to include timestamps of the events")]})

print(response["messages"][-1].content)
```
