# __SLUG__: exploratory analysis

Goal: understand the target, data quality, and how test differs from train. That decides the validation plan.

```python
import pandas as pd

from kagglekit import load_project

project = load_project()
train = project.read_train()
test = project.read_test()
target = project.target_column()
print(train.shape, test.shape, target)
train.head()
```

```python
train[target].describe()
```

```python
features = [column for column in train.columns if column != target]
pd.DataFrame(
    {
        "dtype": train[features].dtypes,
        "unique": train[features].nunique(),
        "train_missing": train[features].isna().mean(),
        "test_missing": test[[c for c in features if c in test]].isna().mean(),
    }
)
```
